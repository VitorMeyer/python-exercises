Calcular a Área de um paralelogramo

```python
print('Bem vindo, esse é um exemplo de como calcular a área de um paralelogramo')
```

```python
base = float(input('Inserir o comprimento da base:'))
```

```python
altura = float(input("Insira a altura:"))
```

```python
areaParalelogramo = base * altura
print("A área do paralelogramo é: ", areaParalelogramo)
```

## Calculadora

```python
numA = float(input("1º Número: "))
```

```python
numB = float(input('2º Número: '))
```

```python
operacao = input("operação: ")
```

```python
if operacao == "+":
    resultado = numA + numB
    print(resultado)
    
elif operacao == "-":
    resultado = numA - numB
    print(resultado)
    
elif operacao == "/" :
    resultado = numA / numB
    print(resultado)
    
elif operacao == "*":
    resultado = numA * numB
    print(resultado)
else:
    print("operação Inválida.")
```
